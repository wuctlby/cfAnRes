```python
import os
from pathlib import Path

def list_files(directory, prefix="", suffix=""):
    files = []
    for root, dirs, filenames in os.walk(directory):
        for filename in filenames:
            if filename.startswith(prefix) and prefix != "":
                pass
            elif prefix == "":
                pass
            else:
                continue
            if filename.endswith(suffix) and suffix != "":
                pass
            elif suffix == "":
                pass
            else:
                continue
            files.append(os.path.join(root, filename))
    return files

def merge_files(input_path, prefix="", suffix="", output_path=""):
    files = list_files(input_path, prefix=prefix, suffix=suffix)
    if not output_path:
        output_path = os.path.join(input_path, "merged.root")
    cmd = f"hadd -f -j 16 {output_path} {' '.join(files)}"
    os.system(cmd)
    print(f"Merged {len(files)} files into {output_path}")

def merge_files_by_file(files, output_path=""):
    if not output_path:
        output_path = os.path.join(os.path.dirname(files[0]), "merged.root")
    cmd = f"hadd -f -j 16 {output_path} {' '.join(files)}"
    print(cmd)
    # os.system(cmd)
    # print(f"Merged {len(files)} files into {output_path}")
    
path = "/home/wuct/MetaData/MC/production/PbPb/test_apass4/k4b"
prefix = ""
suffix = ".root"
files = list_files(path, prefix=prefix, suffix=suffix)
for f in files:
    print(f)
```
