```python
import os
import sys
sys.path.append('/home/wuct/ALICE/reps/cfAnRes/tools/template')
from merge_pdf import collage_png_pages_to_single
from pathlib import Path
import natsort
import re

# ============================================================
# 配置
# ============================================================
base = Path("/home/wuct/MetaData/DATA/OO/apass2/corr/results/fifth/k020_gausPer/etaVariation/CorrelExtract_0d2_1d3_AppDeltaPhi/CorrelationFitResults/outputPathOutput_CorrelationFitting_png")

# 1. 收集 CorrPhiD0 文件 (排除已 merged 的)
corr_files = []
for item in base.iterdir():
    if item.is_file() and item.suffix == ".png" and "CorrPhiD0_PtBinCand" in item.name and "merged" not in item.name:
        corr_files.append(str(item))
corr_sorted = natsort.natsorted(corr_files)

# 2. 收集 hLMtemplate 文件 (排除已 merged 的)
hlm_files = []
for item in base.iterdir():
    if item.is_file() and item.suffix == ".png" and "hLMtemplate_PtBinCand" in item.name and "merged" not in item.name:
        hlm_files.append(str(item))
hlm_sorted = natsort.natsorted(hlm_files)

print(f"CorrPhiD0: {len(corr_sorted)} files")
print(f"hLMtemplate: {len(hlm_sorted)} files")

# 3. 按 PtBinCand 编号配对: [CorrPhiD0_1, hLMtemplate_1, CorrPhiD0_2, hLMtemplate_2, ...]
pt_re = re.compile(r'PtBinCand(\d+)')
pairs_flat = []
for cf, hf in zip(corr_sorted, hlm_sorted):
    cnum = pt_re.search(cf).group(1)
    hnum = pt_re.search(hf).group(1)
    assert cnum == hnum, f"配对失败: PtBinCand{cnum} vs PtBinCand{hnum}"
    pairs_flat.append(cf)
    pairs_flat.append(hf)
    print(f"  Pair PtBinCand{cnum}: CorrPhiD0 + hLMtemplate")

print(f"\n共 {len(pairs_flat)//2} 对, {len(pairs_flat)} 张图")

# 4. 每 4 对 (8 张图) 拼成一张大图: 2 列 × 4 行
pairs_per_merged = 4
imgs_per_merged = pairs_per_merged * 2  # 8

for i in range(0, len(pairs_flat), imgs_per_merged):
    chunk = pairs_flat[i:i + imgs_per_merged]

    # 确定范围
    first = pt_re.search(chunk[0]).group(1)
    last = pt_re.search(chunk[-2]).group(1)  # chunk[-2] 是最后一对的 CorrPhiD0

    outname = base / f"merged_pair_CorrPhiD0_hLMtemplate_PtBinCand{first}to{last}.png"
    print(f"\n合并 PtBinCand{first}-{last}: {len(chunk)} images -> {outname}")

    if len(chunk) < imgs_per_merged:
        collage_png_pages_to_single(
            input_png=chunk,
            output_png=str(outname),
            page_width=1800,
            page_height=1400,   # 2列×4行 → 正方形画布
            cols=2,             # 左 CorrPhiD0, 右 hLMtemplate
            margin=0
        )
    else:
        collage_png_pages_to_single(
            input_png=chunk,
            output_png=str(outname),
            page_width=3600,
            page_height=1400,   # 2列×4行 → 正方形画布
            cols=4,             # 左 CorrPhiD0, 右 hLMtemplate
            margin=0
        )

print("\n完成!")
```

```python
# slice the pdf into sub pdfs
pdf_path = '/home/wuct/MetaData/DATA/OO/apass2/corr/results/fifth/k020/fitprocedure/CorrelExtract_0d8_1d3_Appmass/results_free_woBL/CorrelationFitResults/raw_yields/InvMassVsV2_PtAssoc02to30.pdf'
from pypdf import PdfReader, PdfWriter, Transformation
import fitz  # PyMuPDF

reader = PdfReader(pdf_path)
num_pages = len(reader.pages)
print(f"Total number of pages: {num_pages}")
outpdfs = []
for i in range(num_pages):
    writer = PdfWriter()
    writer.add_page(reader.pages[i])
    output_pdf_path = f"/home/wuct/MetaData/DATA/OO/apass2/corr/results/fifth/k020/fitprocedure/CorrelExtract_0d8_1d3_Appmass/results_free_woBL/CorrelationFitResults/raw_yields/InvMassVsV2_PtAssoc02to30_page_{i+1}.pdf"
    with open(output_pdf_path, "wb") as output_pdf:
        writer.write(output_pdf)
    outpdfs.append(output_pdf_path)

# convert the sliced pdfs to pngs
for pdf in outpdfs:
    doc = fitz.open(pdf)
    for page_num in range(len(doc)):
        page = doc.load_page(page_num)
        pix = page.get_pixmap()
        output_png_path = pdf.replace(".pdf", ".png")
        pix.save(output_png_path)
```
